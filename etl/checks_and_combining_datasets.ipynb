```python
import pandas as pd
import numpy as np

# ============================================================
# 1. LOAD OUTPATIENT PROCESSED DATA
# ============================================================

outpatient_path = r"../data/processed/primary/Outpatient_Claims_Processed.csv"

outpatient = pd.read_csv(outpatient_path, low_memory=False)

print("Outpatient shape:", outpatient.shape)
print("Columns:", len(outpatient.columns))


# ============================================================
# 2. CREATE CLAIM KEY
# ============================================================
# Outpatient has duplicate CLM_ID values because of SEGMENT.
# Therefore CLM_ID alone should NOT be used as the unique key.

outpatient["CLAIM_KEY"] = (
    outpatient["CLM_ID"].astype(str)
    + "_"
    + outpatient["SEGMENT"].astype(str)
)

print("Rows:", len(outpatient))
print("Unique CLAIM_KEY:", outpatient["CLAIM_KEY"].nunique())
print("Duplicate CLAIM_KEY:", outpatient["CLAIM_KEY"].duplicated().sum())


# ============================================================
# 3. DATE FEATURES
# ============================================================

outpatient["CLM_FROM_DT"] = pd.to_datetime(
    outpatient["CLM_FROM_DT"].astype("Int64").astype(str),
    format="%Y%m%d",
    errors="coerce"
)

outpatient["CLM_THRU_DT"] = pd.to_datetime(
    outpatient["CLM_THRU_DT"].astype("Int64").astype(str),
    format="%Y%m%d",
    errors="coerce"
)

outpatient["CLAIM_DURATION_DAYS"] = (
    outpatient["CLM_THRU_DT"] - outpatient["CLM_FROM_DT"]
).dt.days + 1

outpatient["CLAIM_YEAR"] = outpatient["CLM_FROM_DT"].dt.year
outpatient["CLAIM_MONTH"] = outpatient["CLM_FROM_DT"].dt.month


# ============================================================
# 4. IDENTIFY DIAGNOSIS / PROCEDURE / HCPCS COLUMNS
# ============================================================

diagnosis_cols = [
    c for c in outpatient.columns
    if c.startswith("ICD9_DGNS_CD_")
]

# Include admitting diagnosis separately
if "ADMTNG_ICD9_DGNS_CD" in outpatient.columns:
    diagnosis_cols.append("ADMTNG_ICD9_DGNS_CD")

procedure_cols = [
    c for c in outpatient.columns
    if c.startswith("ICD9_PRCDR_CD_")
]

hcpcs_cols = [
    c for c in outpatient.columns
    if c.startswith("HCPCS_CD_")
]

print("Diagnosis columns:", len(diagnosis_cols))
print("Procedure columns:", len(procedure_cols))
print("HCPCS columns:", len(hcpcs_cols))


# ============================================================
# 5. CLAIM-LEVEL COUNTS
# ============================================================

outpatient["DIAGNOSIS_COUNT"] = (
    outpatient[diagnosis_cols]
    .notna()
    .sum(axis=1)
)

outpatient["PROCEDURE_COUNT"] = (
    outpatient[procedure_cols]
    .notna()
    .sum(axis=1)
)

outpatient["HCPCS_COUNT"] = (
    outpatient[hcpcs_cols]
    .notna()
    .sum(axis=1)
)

outpatient["HAS_DIAGNOSIS"] = (
    outpatient["DIAGNOSIS_COUNT"] > 0
).astype(int)

outpatient["HAS_PROCEDURE"] = (
    outpatient["PROCEDURE_COUNT"] > 0
).astype(int)

outpatient["HAS_HCPCS"] = (
    outpatient["HCPCS_COUNT"] > 0
).astype(int)


# ============================================================
# 6. PAYMENT FLAGS
# ============================================================

outpatient["HAS_NEGATIVE_PAYMENT"] = (
    outpatient["CLM_PMT_AMT"] < 0
).astype(int)

outpatient["HAS_PRIMARY_PAYER_PAYMENT"] = (
    outpatient["NCH_PRMRY_PYR_CLM_PD_AMT"] > 0
).astype(int)


# ============================================================
# 7. SEGMENT FEATURES
# ============================================================

outpatient["IS_SEGMENT_2"] = (
    outpatient["SEGMENT"] == 2
).astype(int)

segment1_ids = set(
    outpatient.loc[
        outpatient["SEGMENT"] == 1,
        "CLM_ID"
    ]
)

outpatient["HAS_SEGMENT_1_MATCH"] = (
    outpatient["CLM_ID"].isin(segment1_ids)
).astype(int)


# ============================================================
# 8. TOTAL REIMBURSEMENT
# ============================================================

reimbursement_cols = [
    "CLM_PMT_AMT",
    "NCH_PRMRY_PYR_CLM_PD_AMT",
    "NCH_BENE_BLOOD_DDCTBL_LBLTY_AM",
    "NCH_BENE_PTB_DDCTBL_AMT",
    "NCH_BENE_PTB_COINSRNC_AMT"
]

# Only use columns that actually exist
reimbursement_cols = [
    c for c in reimbursement_cols
    if c in outpatient.columns
]

outpatient["TOTAL_REIMBURSEMENT"] = (
    outpatient[reimbursement_cols]
    .fillna(0)
    .sum(axis=1)
)


# ============================================================
# 9. SELECT CLAIM-LEVEL FEATURES
# ============================================================

outpatient_features = outpatient[
    [
        "CLAIM_KEY",
        "DESYNPUF_ID",
        "CLM_ID",
        "SEGMENT",
        "PRVDR_NUM",

        "CLM_PMT_AMT",
        "NCH_PRMRY_PYR_CLM_PD_AMT",

        "NCH_BENE_BLOOD_DDCTBL_LBLTY_AM",
        "NCH_BENE_PTB_DDCTBL_AMT",
        "NCH_BENE_PTB_COINSRNC_AMT",

        "TOTAL_REIMBURSEMENT",

        "CLM_FROM_DT",
        "CLM_THRU_DT",
        "CLAIM_DURATION_DAYS",
        "CLAIM_YEAR",
        "CLAIM_MONTH",

        "DIAGNOSIS_COUNT",
        "PROCEDURE_COUNT",
        "HCPCS_COUNT",

        "HAS_DIAGNOSIS",
        "HAS_PROCEDURE",
        "HAS_HCPCS",
        "HAS_NEGATIVE_PAYMENT",
        "HAS_PRIMARY_PAYER_PAYMENT",

        "IS_SEGMENT_2",
        "HAS_SEGMENT_1_MATCH"
    ]
].copy()


# ============================================================
# 10. BASIC VALIDATION
# ============================================================

print("\n====================================")
print("OUTPATIENT FEATURE DATASET")
print("====================================")

print("Shape:", outpatient_features.shape)

print("\nDuplicate CLAIM_KEY:")
print(
    outpatient_features["CLAIM_KEY"].duplicated().sum()
)

print("\nMissing values:")
print(
    outpatient_features.isna().sum()
)

print("\nClaim duration:")
print(
    outpatient_features["CLAIM_DURATION_DAYS"].describe()
)

print("\nDiagnosis count:")
print(
    outpatient_features["DIAGNOSIS_COUNT"].describe()
)

print("\nProcedure count:")
print(
    outpatient_features["PROCEDURE_COUNT"].describe()
)

print("\nHCPCS count:")
print(
    outpatient_features["HCPCS_COUNT"].describe()
)

print("\nSegment:")
print(
    outpatient_features["SEGMENT"].value_counts()
)

print("\nNegative payment:")
print(
    outpatient_features["HAS_NEGATIVE_PAYMENT"].value_counts()
)

print("\nPrimary payer:")
print(
    outpatient_features["HAS_PRIMARY_PAYER_PAYMENT"].value_counts()
)
```

```python
outpatient_features_path = (
    r"../data/processed/primary/outpatient_claim_features_for_combining.csv"
)

outpatient_features.to_csv(
    outpatient_features_path,
    index=False
)

print("Saved:", outpatient_features_path)
print("Shape:", outpatient_features.shape)
```
